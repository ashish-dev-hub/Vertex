### Feature engineering

```python
df['experience_gap'] = (df['years_experience'] - df['required_experience'])
```

```python
df['project_internship_score'] = (df['student_num_projects'] +df['student_internships'])
```

```python
df['technical_activity_score'] = (df['student_github_repos'] +df['student_hackathons_participated'])
```
