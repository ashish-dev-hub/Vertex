```python
import joblib
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity 
```

```python
model=joblib.load('recommender.pkl')
tfidf=model['tfidf']
km=model['km']
cluster_role=model['cluster_roles']
jobs=model['jobs']
job_vecs=tfidf.transform(jobs['required_skills'])
```

```python
def recommand(user_skills,top_n=5):
    A=tfidf.transform([user_skills])

    cluster=km.predict(A)[0]
    all_jobs=jobs.assign(similarity=cosine_similarity(A,job_vecs)[0].round(2))
    result=all_jobs[all_jobs.job_title.isin(cluster_role[cluster])]
    if result.empty:
        result=all_jobs
    result = all_jobs[all_jobs.job_title.isin(cluster_role.get(int(cluster), []))]
```

```python
print(recommand('Python,SQL,Pandas'))
```
